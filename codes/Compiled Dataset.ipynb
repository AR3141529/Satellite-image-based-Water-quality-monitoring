```python
import pandas as pd
from functools import reduce
```

```python
csv_files = [
    '/Users/andydascikid/Downloads/Environmental Dataset.csv',
    '/Users/andydascikid/Python Scripts/Home/Yamuna Water Analysis/CSV Files/NDTI_Interpolated.csv',
    '/Users/andydascikid/Python Scripts/Home/Yamuna Water Analysis/CSV Files/DPCC_Interpolated.csv', 
    '/Users/andydascikid/Python Scripts/Home/Yamuna Water Analysis/CSV Files/NDCI_Interpolated.csv', 
    ]
```

```python
dfs = []
for file in csv_files:
    df = pd.read_csv(file)

    df['Date'] = pd.to_datetime(df['Date'])

# 3. Change the format to YYYY-MM-DD
# %Y = 4-digit year, %m = 2-digit month, %d = 2-digit day
    df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
    
    start_date = '2020-01-01'
    end_date = '2025-12-20'

    df = df[((df['Date'] >= start_date) & (df['Date'] <= end_date))]
    # Standardize: Ensure the 'Date' column is converted to datetime 
    # so they match perfectly across all files
    df['Date'] = pd.to_datetime(df['Date'])
    
    dfs.append(df)
```

```python
df_merged = reduce(lambda left, right: pd.merge(left, right, on='Date', how='outer'), dfs)
```

```python
df_merged = df_merged.sort_values(by='Date').reset_index(drop=True)
```

```python
df_merged = df_merged.drop(columns='Unnamed: 0')
```

```python
df_merged = df_merged.rename(columns={
    'Rainfall (mm)' : 'Rainfall [mm]',
    'Air Temp (Celcius)' : 'Air Temperature [C]',
    'Wind Speed (m/s)' : 'Wind Speed [m/s]',
    'Solar Radiation (MJ/m2)' : 'Solar Radiation [MJ/m^2]',
    'Water Surface Temp (C)' : 'Water Surface Temperature [C]',
    'mean_NDTI' : 'NDTI',
    'mean_NDCI' : 'NDCI'
})
```

```python
df_merged.to_csv('Final Dataset.csv')
```
