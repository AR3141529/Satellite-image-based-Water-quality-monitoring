```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/andydascikid/Downloads/DPCC Data.csv')
```

```python
df['Date'] = df['Date'].str.strip()
```

```python
df['parsed_date'] = pd.to_datetime(df['Date'], format='%B, %Y')
```

```python
def get_days_of_month(start_of_month):
    end_of_month = start_of_month + pd.offsets.MonthEnd(0)
    return pd.date_range(start_of_month, end_of_month)
```

```python
df['Daily_Date'] = df['parsed_date'].apply(get_days_of_month)
```

```python
df_expanded = df.explode('Daily_Date')
```

```python
df_expanded = df_expanded.drop(columns=['parsed_date'])
df_expanded = df_expanded.drop(columns='Date')
df_expanded = df_expanded.reset_index(drop=True)
```

```python
cols = ['Daily_Date'] + [col for col in df_expanded.columns if col != 'Daily_Date']
df_expanded = df_expanded[cols]
```

```python
start_date = '2025-12-21'
end_date = '2025-12-31'

df_expanded = df_expanded[~((df_expanded['Daily_Date'] >= start_date) & (df_expanded['Daily_Date'] <= end_date))]
```

```python
df_expanded = df_expanded.rename(columns={
    'Daily_Date' : 'Date'
}).reset_index(drop='True')
```

```python
df_expanded.to_csv('DPCC_Interpolated.csv')
```
