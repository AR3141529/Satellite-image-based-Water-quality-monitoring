```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/andydascikid/Downloads/Mean NDCI.csv')
```

```python
df['Date'] = pd.to_datetime(df['system:time_start'], format='%b %d, %Y')

df = df.drop(columns='system:time_start')

new_order = ['Date', 'mean_NDCI']

df = df[new_order]
```

```python
df.set_index('Date', inplace=True)

df = df.groupby(level=0).mean()

df_resampled = df.resample('D').asfreq()

df_resampled['mean_NDCI'] = df_resampled['mean_NDCI'].interpolate(method='linear')

df_final = df_resampled.reset_index()

df_final = df_final.round(5)
```

```python
df_final.to_csv('NDCI_Interpolated.csv', index=False)
```
